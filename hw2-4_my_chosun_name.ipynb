```python
month_names = ['쌍', '쇠', '복', '돌', '팽', '육', '쌍', '개', '칠', '갑', '삼', '방']
day_names = ['봉', '구', '욕', '포', '똥', '삼', '식', '석', '놈', '님', '년', '돌', '단', '득', '방', '질', '장', '걸', '래', '룡', '동', '순', '자', '박', '창', '언', '것', '포', '만', '단', '국']

def get_my_chosun_name(family_name, month, day):
    first = month_names[month - 1]
    second = day_names[day - 1]
    return f"당신의 조선시대 이름은 {family_name}{first}{second} 입니다."

print(get_my_chosun_name('김', 8, 16))
```
